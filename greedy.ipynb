Makes test cases:

```python
import pandas as pd 
import numpy as np 
import math 
import random as rd
import warnings
warnings.filterwarnings('ignore')

def genRec(length): #generate the recurrence for the test case
    rec = ''
    for i in range(length):
        rec += f'{rd.randint(0,1)}'
    return rec

def genSeq(slength, rlength): #generate the full sequence for the test case
    seq = ''
    rec = genRec(rlength)
    for i in range(slength):
        seq += rec
    return seq, rec, 0, 0

tests = pd.DataFrame(columns=['Test Case', 'Reccurence', 'Solution', 'Works?']) #initialize the dataframe
for k in range(10): #generate the test cases
    tests.loc[k] = genSeq(2,10004571)
```

Decode and test:

```python
testList = tests['Test Case'].to_list() #send each test case to a list

def decode(sequence):
    n = len(sequence) #get length
    factors = []
    for j in range(n-2): #find factors
        if n % (j+2) == 0:
            factors += [n/(j+2)]
    
    mFact = max(factors) #pick the largest factor
    rec = sequence[int(n - mFact + 2):n] #isolate the recurrence
    return rec, int(n-mFact+2), n

l = 0
for sequence in testList:
    tests.loc[l,'Solution'], repets, n = decode(sequence)
    
    if str(tests.loc[l,'Test Case'])[repets:n] == tests.loc[l,'Solution']: #checks our work
        tests.loc[l,'Works?'] = 1
    l += 1

display(tests)
```
