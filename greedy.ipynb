Makes test cases:

```python
import pandas as pd 
import numpy as np 
import math 
import random as rd

def genRec(length):
    rec = ''
    for i in range(length):
        rec += f'{rd.randint(0,1)}'
    return rec

def genSeq(slength, rlength):
    seq = ''
    rec = genRec(rlength)
    for i in range(slength):
        seq += rec
    return seq, rec, 0, 0

tests = pd.DataFrame(columns=['Test Case', 'Reccurence', 'Solution', 'Works?'])
for k in range(15):
    tests.loc[k] = genSeq(5,137)
```

Decode and test:

```python
testList = tests['Test Case'].to_list()

def decode(sequence):
    n = len(sequence)
    factors = []
    for j in range(n-2):
        if n % (j+2) == 0:
            factors += [n/(j+2)]
    
    mFact = max(factors)
    rec = sequence[int(n - mFact + 2):n]
    return rec, int(n-mFact+2), n

l = 0
for sequence in testList:
    tests.loc[l,'Solution'], repets, n = decode(sequence)
    if str(tests.loc[l,'Test Case'])[repets:n] == tests.loc[l,'Solution']: #checks our work
        tests.loc[l,'Works?'] = 1
    l += 1

display(tests)
```
